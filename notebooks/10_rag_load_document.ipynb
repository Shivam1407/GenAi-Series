#### FOR PDF 

```python
from langchain_community.document_loaders import PyPDFLoader

# 1. Initialize the loader
loader = PyPDFLoader("../data/Electricity (Prashant kirad).pdf")

# 2. Load the document
docs = loader.load()

# 3. Accessing Data
# print(len(docs)) # Will output the total number of pages in the PDF (e.g., 9)
print(docs[0].page_content) # Text from Page 1
# print(docs[0].metadata) # Metadata (like source path and page number)
```

#### csv

```python
from langchain_community.document_loaders import CSVLoader

# 1. Initialize the loader
loader = CSVLoader("../data/Daily Goal.csv")

# 2. Load the document
docs = loader.load()

# 3. Accessing Data
print(len(docs)) # Outputs the number of rows in the CSV (e.g., 30000)
# View the first 5 records
for doc in docs[:5]:
    print(doc.page_content)
```

#### for web page loader

```python
from langchain_community.document_loaders import WebBaseLoader

# 1. Initialize the loader with a URL
loader = WebBaseLoader("https://www.techsimplus.com/some-course-page")

# 2. Load the page
docs = loader.load()

# 3. Accessing Data
print(docs[0].page_content) # The scraped text
print(docs[0].metadata['title']) # The HTML title of the webpage
```

#### for wikipidea 

```python
from langchain_community.document_loaders import WikipediaLoader

# 1. Initialize the loader with a search query. 
# It is highly recommended to set load_max_docs to limit execution time.
loader = WikipediaLoader(query="Generative AI", load_max_docs=2)

# 2. Load the articles
docs = loader.load()

# 3. Accessing Data
print(len(docs)) # Outputs 2
print(docs[0].page_content) # First retrieved article's text
```
