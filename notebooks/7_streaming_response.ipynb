```python


from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI
llm = ChatGoogleGenerativeAI(model = "gemini-3.5-flash" , streaming = True)

questions = input("enter your prompt: ")

response = llm.stream(questions)

for chunk in response :
    print(chunk.text,end="")

```
