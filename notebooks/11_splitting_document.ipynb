```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

# --- 1. Load the Documents ---
# Text File
text_loader = TextLoader("../data/speech.txt")
text_docs = text_loader.load()

# PDF File
pdf_loader = PyPDFLoader("../data/medical_report.pdf")
pdf_docs = pdf_loader.load()

# Check initial page count of PDF
print(f"Original PDF Pages: {len(pdf_docs)}") # Outputs: 9

# --- 2. Initialize the Splitter ---
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, 
    chunk_overlap=200
)

# --- 3. Split the Documents ---
text_chunks = splitter.split_documents(text_docs)
pdf_chunks = splitter.split_documents(pdf_docs)

# --- 4. Check Results ---
print(f"Total Text Chunks Created: {len(text_chunks)}") 

# The 9-page PDF is now split into multiple smaller Document chunks based on the chunk_size
print(f"Total PDF Chunks Created: {len(pdf_chunks)}") # Outputs: 26

# If we decrease the chunk_size, the number of chunks increases
# (e.g., chunk_size=500, chunk_overlap=100 would result in ~48 chunks for the same PDF)
```
