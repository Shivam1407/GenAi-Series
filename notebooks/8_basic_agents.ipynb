```python
from dotenv import load_dotenv
load_dotenv()

from langchain_core.tools import tool

@tool
def add_numbers(a:int, b:int):
     """
        it will return the sum of two numbers
        Args:
            a: number one
            b: number two
     """
     return a+b


@tool
def multiply_number(a:int, b:int):
     """
        it will return the product of two numbers
        Args:
            a: number one
            b: number two
     """
     return a*b
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

agent = create_agent(
    model= llm,
    tools= [add_numbers, multiply_number],
    system_prompt="You are a maths teacher and always use tools caluclaion"
)
query = input("enter the query")
response = agent.invoke({
    "messages":[
        {"role":"user","content":query}
    ]
})

final_answer = response["messages"][-1].content

print(final_answer)
```
