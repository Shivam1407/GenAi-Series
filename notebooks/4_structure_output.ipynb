```python
from dotenv import load_dotenv
load_dotenv()

from langchain_google_genai import ChatGoogleGenerativeAI

from pydantic import BaseModel, Field

# Initialize LLM
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

# 1. Define the Expected Structure using Pydantic
class ResponseStructure(BaseModel):
    name: str = Field(description="Complete name of the user")
    email: str = Field(description="Email address of the user")
    age: int = Field(description="Age of the user")

# 2. Bind the structure to the LLM
structured_llm = llm.with_structured_output(ResponseStructure)

# 3. Create the input text and invoke
text = """
Hello my name is Pratik Mishra. 
Email is pratik29mishra@gmail.com 
and my age is 29.
"""

response = structured_llm.invoke(f"Please give me only name, email and age from this text: {text}")

# 4. Access the outputs programmatically
print(response.name)           # Outputs: Pratik Mishra
print(response.age)            # Outputs: 29
print(response.model_dump())   # Outputs a Python Dictionary: {'name': 'Pratik Mishra', 'email': 'pratik29mishra@gmail.com', 'age': 29}
```

```python
from dotenv import load_dotenv
load_dotenv()
from typing import List
from pydantic import BaseModel, Field
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")
class Movies(BaseModel):
    Name: str = Field(description="It contains the name of the movie")
    Year: int = Field(description="it contains the year in which movies was releasesd")

class AllMovies(BaseModel):
    MovieDetail: List[Movies]

structure_llm = llm.with_structured_output(AllMovies)
res = structure_llm.invoke("List me 3 romantic movies")
print(res)
```
