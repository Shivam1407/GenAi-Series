#### Field Validation


```python
from pydantic import BaseModel , Field
from typing import List, Annotated

class Product(BaseModel):
    title: str = Field(max_length=100)
    des : str = Field(min_length=20)
    price : int = Field(gt=0,lt=1000000)
```

```python
product = Product(title="Samsung s24",des="maka laadle ye wo hai jo hona chaiye tere pass iphone me kya rkha hai",price=124000)
print(product.title,"\n",product.des,"\n",product.price)
```

```python
class Order(BaseModel):
    id:str = Field(min_length=10,max_length=100)
    product:Product

order =Order(id="order12345",product=product)
```

```python
print(order)
print(order.product.price)
```

```python
class Carts(BaseModel):
    id:str = Field(min_length=3,max_length=20)
    orders:List[Order]
```

```python
cart= Carts(id="id01",orders=[order])
print(cart)
print(cart.orders[0].product.price)
```

```python
CommonValidator = Annotated[List[Product], Field(default=[])]

class Order(BaseModel):
    # Now we can just use the custom type instead of rewriting the rules
    products: CommonValidator



# Creating an order with products
filled_order = Order(product=[product])
print(filled_order.products)
```
