#### prompt templates - dynamic prompts


```python

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

prompts = ChatPromptTemplate.from_messages([
    {
        "role": "system",
        "content": "You are a translator and translate the input to {language}."
    },
    {
        "role": "user",
        "content": "{query}"
    },
])

final_prompts = prompts.format_messages(
    language="Hinglish ?",
    query="I love myself only, got it."
)

res = llm.invoke(final_prompts)

print(res.content)
```

####we can directly send one runnables output as another runnables input

```python

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

prompts = ChatPromptTemplate.from_messages([
    {
        "role": "system",
        "content": "You are a translator and translate the input to {language}."
    },
    {
        "role": "user",
        "content": "{query}"
    },
])

res =  llm.invoke(prompts.invoke({"language":"Hinglish","query": "Being a positive deviation solution"}))

print(res.content)
```

#chain

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers  import StrOutputParser

llm = ChatGoogleGenerativeAI(model="gemini-3.5-flash")

prompts = ChatPromptTemplate.from_messages([
    {"role":"system","content":"you are a translater and translate the input to {language}"},
    {"role":"user","content":"{query}"},
])

out_parser= StrOutputParser()

def tranform_case(result:str)->str:
    return result.upper()

chain = prompts | llm | out_parser | tranform_case
txt = input("enter the text right here :")
res = chain.invoke({
    "language":"Hinglish",
    "query": "{txt}"
})

print(res)
```
